A soil-moisture sensor sometimes glitches and logs a negative value, which is physically impossible. A simple fix: replace every negative reading with the mean of all the genuinely non-negative readings in the same batch.

Write replace_negatives_with_mean(arr) that takes a 1D NumPy array arr and returns a new array (do not modify arr in place) where every negative entry has been replaced by the mean of the non-negative entries of arr. Do not use a Python loop; use boolean indexing.

Hint: if you're not sure arr is already a float array, cast it up front with arr.astype(float) — otherwise assigning a mean into an integer array will silently truncate it.

For example,

replace_negatives_with_mean(np.array([4, -1, 6, -3, 8]))
returns array([4., 6., 6., 6., 8.]) (the non-negative values are 4, 6, 8, whose mean is 6).

```python
 import numpy as np
 
 
 def replace_negatives_with_mean(arr):
   arr=arr.astype(float)
   mean = np.mean(arr[arr>=0])
   arr[arr<0]=mean
   return arr
```

A teacher wants to convert a class's raw exam scores into z-scores, so that scores can be compared across different tests. The z-score of a value x is (x - mean) / std, where std is the population (biased) standard deviation of the whole batch (i.e. ddof=0, NumPy's default).

Write z_scores(data) that takes a 1D array-like data of raw scores and returns a NumPy array of the corresponding z-scores.

For example,

z_scores(np.array([2, 4, 4, 4, 5, 5, 7, 9]))
returns array([-1.5, -0.5, -0.5, -0.5, 0., 0., 1., 2.]).

```python
 import numpy as np
 
 
 def z_scores(data):
   return (data-np.mean(data))/np.std(data)
```

To process a refund, a shopkeeper is going through a list amounts of transaction amounts, in whole rupees (so every value is an integer — comparing them for exact equality is safe here, unlike with amounts that could have paise/decimal parts).

She wants to know how many pairs of positions (i, j) in the list, with i < j (so each pair of entries is counted once, not twice, and no entry is paired with itself), have amounts[i] + amounts[j] == target.

Write count_pairs(amounts, target) returning how many such pairs of positions exist.

For example,
count_pairs([1, 5, 3, 5, 2], 6)
returns 2. (The two qualifying pairs of positions are (0, 1), since amounts[0] + amounts[1] = 1 + 5 = 6, and (0, 3), since amounts[0] + amounts[3] = 1 + 5 = 6 — position 3 holds a second 5, distinct from position 1.)

```python
 def count_pairs(amounts, target):
     count=0
     for i in range(len(amounts)):
       for j in range(i+1,len(amounts)):
         if amounts[i]+amounts[j]==target:
           count=count+1
     return count
```

Write rotate_list(seq, k) that returns a new list with the elements of seq rotated cyclically to the right by k positions.

Positive k: rotate right by k, wrapping the trailing elements to the front.
Negative k: rotate left by |k|, wrapping the leading elements to the back.
If |k| >= len(seq), use the effective rotation k mod n.
seq must not be modified. Return [] if seq is empty.
For example,
rotate_list([10, 20, 30, 40, 50, 60], 2)
returns [50, 60, 10, 20, 30, 40].

```python
 def rotate_list(seq, k):
   if(len(seq)==0):
     return []
   k=k%len(seq)
   return seq[len(seq)-k:]+seq[:len(seq)-k]
```

A factory floor has several pressure sensors, each with its own valid operating range — they're different makes of sensor, calibrated differently. Readings are logged in a 2D NumPy array readings of shape (num_sensors, num_readings), one row per sensor. Each sensor's valid range is given by two 1D arrays low and high, each of length num_sensors, where low[i] and high[i] are the valid bounds for sensor i.

Write clip_per_sensor(readings, low, high) that returns a new array of the same shape as readings, in which every entry in row i has been clipped to [low[i], high[i]] — using broadcasting, not a Python loop over sensors.

Hint: low and high are 1D, one value per row of readings. You'll need to turn each into a column before it can broadcast correctly against a 2D array — low[:, np.newaxis] does this.

For example,
readings = np.array([[-5,  3, 12],
                      [ 8, -1, 20],
                      [ 2, 25, -10]])
low  = np.array([0, -2, 5])
high = np.array([10, 15, 15])

clip_per_sensor(readings, low, high)
returns
array([[ 0.,  3., 10.],
       [ 8., -1., 15.],
       [ 5., 15.,  5.]])


```python
 import numpy as np
 
 
 def clip_per_sensor(readings, low, high):
     return np.clip(readings,low[:,np.newaxis],high[:,np.newaxis])
```

A classic Monte Carlo trick: estimate the value of π by throwing random points at a square dartboard and seeing how many land inside the inscribed circle.

Complete estimate_pi(n_points, seed), started for you below — the seed is already set, so you don't need to call np.random.seed yourself:
def estimate_pi(n_points, seed):
    np.random.seed(seed)
    # your code here: generate n_points random 2D points, with x and y
    # coordinates independently and uniformly distributed in [-1, 1]
    # (np.random.uniform is fine here), compute the fraction that land
    # inside the unit circle (x**2 + y**2 <= 1), and return the pi
    # estimate 4 * that fraction, as a Python float
For example,
estimate_pi(500000, 7)
returns approximately 3.14159 (e.g. 3.142712).

```python
 import numpy as np
 
 
 def estimate_pi(n_points, seed):
     np.random.seed(seed)
     pts=np.random.uniform(-1,1,(n_points,2))
     inside_pts=(pts[:,0]**2+pts[:,1]**2) <=1
     return float(4*np.mean(inside_pts))
```
