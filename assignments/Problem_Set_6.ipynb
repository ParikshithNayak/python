A cricket selector wants a batsman's best stretch of consecutive innings in which every score beat the one before it. That is his longest improving streak. Write longest_increasing_run(scores) that returns the length of this streak. If the batsman has not played any innings yet, return 0. A single innings is a streak of length 1. Example: longest_increasing_run([50, 60, 40, 45, 55, 70, 90, 30]) returns 5, because the stretch 40, 45, 55, 70, 90 has five innings.

```python
def longest_increasing_run(scores):
    if len(scores) == 0:
      return 0
    streak=1
    max_streak=1
    for i in range(1,len(scores)):
      if scores[i-1] < scores[i]:
        streak+=1
        if max_streak < streak:
          max_streak+=1
      else:
        streak=1
    return max_streak
```

A surveyor's elevation readings run along a ridge in the Himalayas. A reading counts as a peak when it is strictly higher than both of its immediate neighbours. The first and last readings have only one neighbour each, so they only need to beat that one. A list with a single reading is its own peak. Write count_peaks(elevations) that returns how many peaks the readings contain. Return 0 for an empty list. Example: count_peaks([1, 3, 2, 4, 1, 5]) returns 3, from the readings 3, 4 and the final 5.

```python
def count_peaks(elevations):
  if len(elevations)==0:
    return 0
  elif len(elevations)==1:
    return 1
  else:
    peaks=0
    for i in range(len(elevations)):
      if i==0:
        if elevations[i]>elevations[i+1]:
          peaks+=1
      elif i==len(elevations)-1:
        if elevations[i-1]<elevations[i]:
          peaks+=1
      else:
        if elevations[i-1]<elevations[i] and elevations[i] >elevations[i+1]:
          peaks+=1
    return peaks
        
        
```

Write merge_sorted(list1, list2) that merges two lists, each already sorted in increasing order, into a single sorted list. Do not use the built-in sorted() or the .sort() method. Walk the two lists together with a while loop instead, taking the smaller front element each time. Neither input list may be modified. Example: merge_sorted([1, 3, 5], [2, 4, 6]) returns [1, 2, 3, 4, 5, 6]


```python
def merge_sorted(list1, list2):
  result = []
  i = 0
  j = 0
  while i < len(list1) and j < len(list2):
    if list1[i] <= list2[j]:
      result.append(list1[i])
      i += 1
    else:
      result.append(list2[j])
      j += 1
  while i < len(list1):
      result.append(list1[i])
      i += 1
  while j < len(list2):
      result.append(list2[j])
      j += 1
  return result
```

Write time_cdf(times), where times is a non-empty NumPy array of non-negative integers. Let max_time = int(np.max(times)). Return a NumPy array cdf of length max_time + 1 where cdf[x] is the empirical probability that a value drawn from times is less than or equal to x, for every integer x from 0 to max_time. Values that never occur in times must still have an entry in cdf. Example: time_cdf(np.array([0, 2, 2, 3, 5])) returns array([0.2, 0.2, 0.6, 0.8, 0.8, 1.0])

```python
import numpy as np


def time_cdf(times):
    max_time = int(np.max(times))
    cdf=np.zeros(max_time+1)
    for x in range(max_time+1):
      cdf[x]=np.mean(times<=x)
    return cdf
```

Write initials(full_name) that turns a person's full name into their initials. Take the first letter of every word, put it in upper case, and separate the letters with full stops. The result also ends with a full stop. Extra spaces between words, or at either end of the name, must not affect the answer. Splitting a string with .split() and no argument handles that for you. Return the empty string when the name has no words at all. Example: initials("Mohandas Karamchand Gandhi") returns "M.K.G."

```python
def initials(full_name):
    names = full_name.split()

    if len(names) == 0:
        return ''

    for i in range(len(names)):
        names[i] = names[i][0].upper()

    return '.'.join(names) + '.'
```

Count how often each word appears in a piece of text. Write word_counts(text) that returns a dictionary mapping each word to the number of times it occurs. Before counting, prepare the text as follows.
put the whole text into lower case, so that The and the count together
delete every punctuation character, so that cat, counts as cat
split on whitespace to get the words
Deleting the punctuation is easiest with the string module. Building a translation table from dict.fromkeys(string.punctuation) maps every punctuation character to None, which tells translate to drop it. Note that punctuation is deleted and not turned into a space, so don't becomes dont. Return an empty dictionary for empty text. Example: word_counts("The cat, the hat; the CAT!") returns {'the': 3, 'cat': 2, 'hat': 1}

```python
import string


def word_counts(text):
  if len(text) ==0:
    return {}
  text = text.translate(str.maketrans(dict.fromkeys(string.punctuation))).lower().split()
  word_counts={}
  for word in text:
    if word in word_counts:
      word_counts[word]+=1
    else:
      word_counts[word]=1
  return word_counts
```

A phone book arrives as a list of strings, one entry per string, with the name and the number separated by a colon. The spacing around the colon is not reliable. Write parse_phonebook(lines) that returns a dictionary mapping each name to its number as an integer. Strip the stray spaces from both halves. You may assume every line holds exactly one colon. Return an empty dictionary for an empty list. Example: parse_phonebook(['mom : 1234', 'dad:5678']) returns {'mom': 1234, 'dad': 5678}


```python
def parse_phonebook(lines):
  phone_book={}
  for line in lines:
    line_arr=line.split(':')
    phone_book.update({line_arr[0].strip() : int(line_arr[1].strip())})
  return phone_book
```

Write two functions that are the reverse of each other. save_lines(path, lines) takes a list of strings and writes them to the file at path, one string per line. Every line in the file ends with a newline character, including the last one. load_lines(path) reads that file back and returns the list of strings without the newline characters. Open both files with the with statement so that they are closed for you. You may assume the strings hold no leading or trailing spaces of their own. Example: after save_lines("notes.txt", ["alpha", "beta"]) the file holds "alpha\nbeta\n", and load_lines("notes.txt") returns ['alpha', 'beta']

```python
def save_lines(path, lines):
    with open(path, 'w') as f:
        for line in lines:
            f.write(line + '\n')


def load_lines(path):
    with open(path) as f:
        result = []
        for line in f:
            result.append(line.rstrip('\n'))
        return result
```

Write copy_alternate(src_path, dst_path) that reads the file at src_path and writes every alternate line of it to the file at dst_path, starting with the very first line. So lines 1, 3, 5 and so on are kept and the rest are dropped. Return the number of lines written. Read the whole file with .readlines() and write the kept lines with .writelines(). Remember that .readlines() leaves the newline character on the end of each line, which is exactly what .writelines() needs, because it does not add one of its own. Example: if the source file holds one, two, three, four, five on five lines, then the destination file ends up holding one, three, five, and the function returns 3

```python
def copy_alternate(src_path, dst_path):
    with open(src_path) as src:
        lines = src.readlines()

    kept = lines[::2]

    with open(dst_path, 'w') as dst:
        dst.writelines(kept)

    return len(kept)
```

a column area
a column density, holding the population divided by the area
Build it by turning each dictionary into a pd.Series and handing both to pd.DataFrame. Example: for {'Bihar': 104099452} and {'Bihar': 94163} the row for Bihar holds a population of 104099452, an area of 94163 and a density of about 1105.5

```python
import pandas as pd


def state_table(population, area):
  population=pd.Series(population)
  area=pd.Series(area)
  return pd.DataFrame({"population":population,"area":area,"density":population/area})
```

A DataFrame of sensor readings has gaps in it. A reading that was never recorded shows up as NaN, the special floating point value pandas uses for missing data. Write two functions. count_missing(data, column) returns how many entries of column are missing. Ask the column for .isnull(), which gives a column of True and False, and then sum it. Summing a column of booleans counts the True entries. fill_missing(data, column, value) returns a new DataFrame in which every missing entry of column has been replaced by value. The DataFrame handed in must not be changed, so take a .copy() first and then use .fillna() on the column. Example: for a reading column holding 1.0, NaN, 3.0, count_missing returns 1 and filling with 0.0 gives 1.0, 0.0, 3.0

```python
import pandas as pd



def count_missing(data, column):
    return data[column].isnull().sum()


def fill_missing(data, column, value):
    result = data.copy()
    result[column] = result[column].fillna(value)
    return result
```
