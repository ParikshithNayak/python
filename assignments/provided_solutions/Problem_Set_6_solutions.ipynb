### Question 1. Increasing ODI runs

(2 marks)

A cricket selector wants a batsman's best stretch of consecutive innings in
which every score beat the one before it. That is his longest improving streak.

Write `longest_increasing_run(scores)` that returns the length of this streak.
If the batsman has not played any innings yet, return `0`. A single innings is
a streak of length `1`.

Example:

`longest_increasing_run([50, 60, 40, 45, 55, 70, 90, 30])` returns `5`,
because the stretch `40, 45, 55, 70, 90` has five innings.

```python
def longest_increasing_run(scores):
    if not scores:
        return 0
    best = 1
    current = 1
    for i in range(1, len(scores)):
        if scores[i] > scores[i - 1]:
            current = current + 1
        else:
            current = 1
        if current > best:
            best = current
    return best
```

```python
# Test cases

from hamcrest import assert_that, equal_to
assert_that(longest_increasing_run([50, 60, 40, 45, 55, 70, 90, 30]), equal_to(5))
assert_that(longest_increasing_run([]), equal_to(0))
assert_that(longest_increasing_run([7]), equal_to(1))

from hamcrest import assert_that, equal_to
assert_that(longest_increasing_run([1, 2, 3, 4, 5]), equal_to(5))
assert_that(longest_increasing_run([5, 4, 3, 2, 1]), equal_to(1))
assert_that(longest_increasing_run([2, 2, 2]), equal_to(1))
assert_that(longest_increasing_run([3, 9, 1, 2]), equal_to(2))

print('Question 1: all test cases passed')
```

### Question 2. Counting peaks in the Himalayas

(2 marks)

A surveyor's elevation readings run along a ridge in the Himalayas. A reading
counts as a peak when it is strictly higher than both of its immediate
neighbours. The first and last readings have only one neighbour each, so they
only need to beat that one. A list with a single reading is its own peak.

Write `count_peaks(elevations)` that returns how many peaks the readings
contain. Return `0` for an empty list.

Example:

`count_peaks([1, 3, 2, 4, 1, 5])` returns `3`, from the readings `3`, `4` and
the final `5`.

```python
def count_peaks(elevations):
    n = len(elevations)
    if n == 0:
        return 0
    if n == 1:
        return 1
    count = 0
    for i in range(n):
        if i == 0:
            if elevations[i] > elevations[i + 1]:
                count = count + 1
        elif i == n - 1:
            if elevations[i] > elevations[i - 1]:
                count = count + 1
        else:
            if elevations[i] > elevations[i - 1] and elevations[i] > elevations[i + 1]:
                count = count + 1
    return count
```

```python
# Test cases

from hamcrest import assert_that, equal_to
assert_that(count_peaks([1, 3, 2, 4, 1, 5]), equal_to(3))
assert_that(count_peaks([]), equal_to(0))
assert_that(count_peaks([9]), equal_to(1))

from hamcrest import assert_that, equal_to
assert_that(count_peaks([1, 2, 3]), equal_to(1))
assert_that(count_peaks([3, 2, 1]), equal_to(1))
assert_that(count_peaks([2, 2, 2]), equal_to(0))
assert_that(count_peaks([5, 1, 5]), equal_to(2))

print('Question 2: all test cases passed')
```

### Question 3. Merging two sorted lists

(2 marks)

Write `merge_sorted(list1, list2)` that merges two lists, each already sorted
in increasing order, into a single sorted list.

Do not use the built-in `sorted()` or the `.sort()` method. Walk the two lists
together with a `while` loop instead, taking the smaller front element each
time.

Neither input list may be modified.

Example:

`merge_sorted([1, 3, 5], [2, 4, 6])` returns `[1, 2, 3, 4, 5, 6]`

```python
def merge_sorted(list1, list2):
    result = []
    i = 0
    j = 0
    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            result.append(list1[i])
            i = i + 1
        else:
            result.append(list2[j])
            j = j + 1
    while i < len(list1):
        result.append(list1[i])
        i = i + 1
    while j < len(list2):
        result.append(list2[j])
        j = j + 1
    return result
```

```python
# Test cases

from hamcrest import assert_that, equal_to
assert_that(merge_sorted([1, 3, 5], [2, 4, 6]), equal_to([1, 2, 3, 4, 5, 6]))
assert_that(merge_sorted([], [2, 4]), equal_to([2, 4]))
assert_that(merge_sorted([1, 2], []), equal_to([1, 2]))
assert_that(merge_sorted([], []), equal_to([]))

from hamcrest import assert_that, equal_to
assert_that(merge_sorted([1, 1, 4], [1, 3]), equal_to([1, 1, 1, 3, 4]))
assert_that(merge_sorted([10], [1, 2, 3]), equal_to([1, 2, 3, 10]))

a = [1, 5]
b = [2, 3]
merge_sorted(a, b)
assert_that(a, equal_to([1, 5]))
assert_that(b, equal_to([2, 3]))

print('Question 3: all test cases passed')
```

### Question 4. Empirical CDF

(2 marks)

Write `time_cdf(times)`, where `times` is a non-empty NumPy array of
non-negative integers.

Let `max_time = int(np.max(times))`. Return a NumPy array `cdf` of length
`max_time + 1` where `cdf[x]` is the empirical probability that a value drawn
from `times` is less than or equal to `x`, for every integer `x` from `0` to
`max_time`.

Values that never occur in `times` must still have an entry in `cdf`.

Example:

`time_cdf(np.array([0, 2, 2, 3, 5]))` returns
`array([0.2, 0.2, 0.6, 0.8, 0.8, 1.0])`

```python
import numpy as np


def time_cdf(times):
    times = np.asarray(times)
    max_time = int(np.max(times))
    counts = np.zeros(max_time + 1)
    for t in times:
        counts[t] = counts[t] + 1
    return np.cumsum(counts) / len(times)
```

```python
# Test cases

from hamcrest import assert_that, contains_exactly, close_to, has_length
import numpy as np

result = time_cdf(np.array([0, 2, 2, 3, 5]))
assert_that(result, has_length(6))
assert_that(list(result), contains_exactly(
    close_to(0.2, 1e-9), close_to(0.2, 1e-9), close_to(0.6, 1e-9),
    close_to(0.8, 1e-9), close_to(0.8, 1e-9), close_to(1.0, 1e-9)))

from hamcrest import assert_that, contains_exactly, close_to, has_length
import numpy as np

result = time_cdf(np.array([3]))
assert_that(result, has_length(4))
assert_that(list(result), contains_exactly(
    close_to(0.0, 1e-9), close_to(0.0, 1e-9),
    close_to(0.0, 1e-9), close_to(1.0, 1e-9)))

result = time_cdf(np.array([0, 0, 1]))
assert_that(list(result), contains_exactly(
    close_to(2.0 / 3.0, 1e-9), close_to(1.0, 1e-9)))

print('Question 4: all test cases passed')
```

### Question 5. Initials from a full name

(2 marks)

Write `initials(full_name)` that turns a person's full name into their
initials.

Take the first letter of every word, put it in upper case, and separate the
letters with full stops. The result also ends with a full stop.

Extra spaces between words, or at either end of the name, must not affect the
answer. Splitting a string with `.split()` and no argument handles that for
you. Return the empty string when the name has no words at all.

Example:

`initials("Mohandas Karamchand Gandhi")` returns `"M.K.G."`

```python
def initials(full_name):
    letters = []
    for part in full_name.split():
        letters.append(part[0].upper())
    if len(letters) == 0:
        return ""
    return ".".join(letters) + "."
```

```python
# Test cases

from hamcrest import assert_that, equal_to
assert_that(initials("Mohandas Karamchand Gandhi"), equal_to("M.K.G."))
assert_that(initials("Ram"), equal_to("R."))
assert_that(initials(""), equal_to(""))

from hamcrest import assert_that, equal_to
assert_that(initials("  sarojini   naidu "), equal_to("S.N."))
assert_that(initials("   "), equal_to(""))
assert_that(initials("a b c d"), equal_to("A.B.C.D."))

print('Question 5: all test cases passed')
```

### Question 6. Masking an account number

(2 marks)

A bank statement must never print a full account number. Only the last four
digits may be shown, and every digit before them is replaced by the letter
`X`.

Write `mask_account(number)`, where `number` is a string of digits. Return the
masked string. If the number is four digits or shorter there is nothing to
hide, so return it unchanged.

Build the mask with string repetition and slicing rather than a loop.

Example:

`mask_account("123456789012")` returns `"XXXXXXXX9012"`

```python
def mask_account(number):
    if len(number) <= 4:
        return number
    return "X" * (len(number) - 4) + number[-4:]
```

```python
# Test cases

from hamcrest import assert_that, equal_to
assert_that(mask_account("123456789012"), equal_to("XXXXXXXX9012"))
assert_that(mask_account("56789012"), equal_to("XXXX9012"))

from hamcrest import assert_that, equal_to
assert_that(mask_account("9012"), equal_to("9012"))
assert_that(mask_account("12"), equal_to("12"))
assert_that(mask_account(""), equal_to(""))
assert_that(mask_account("12345"), equal_to("X2345"))

print('Question 6: all test cases passed')
```

### Question 8. Counting the words in a paragraph

(3 marks)

Count how often each word appears in a piece of text.

Write `word_counts(text)` that returns a dictionary mapping each word to the
number of times it occurs. Before counting, prepare the text as follows.

- put the whole text into lower case, so that `The` and `the` count together
- delete every punctuation character, so that `cat,` counts as `cat`
- split on whitespace to get the words

Deleting the punctuation is easiest with the `string` module. Building a
translation table from `dict.fromkeys(string.punctuation)` maps every
punctuation character to `None`, which tells `translate` to drop it.

Note that punctuation is deleted and not turned into a space, so `don't`
becomes `dont`.

Return an empty dictionary for empty text.

Example:

`word_counts("The cat, the hat; the CAT!")` returns
`{'the': 3, 'cat': 2, 'hat': 1}`

```python
import string


def word_counts(text):
    table = dict.fromkeys(string.punctuation)
    cleaned = text.lower().translate(str.maketrans(table))
    counts = {}
    for word in cleaned.split():
        if word in counts:
            counts[word] = counts[word] + 1
        else:
            counts[word] = 1
    return counts
```

```python
# Test cases

from hamcrest import assert_that, equal_to
assert_that(word_counts("The cat, the hat; the CAT!"), equal_to({"the": 3, "cat": 2, "hat": 1}))
assert_that(word_counts(""), equal_to({}))

from hamcrest import assert_that, equal_to
assert_that(word_counts("Don't stop; don't!"), equal_to({"dont": 2, "stop": 1}))
assert_that(word_counts("   spaced    out   "), equal_to({"spaced": 1, "out": 1}))
assert_that(word_counts("Hello, hello. HELLO!"), equal_to({"hello": 3}))

print('Question 8: all test cases passed')
```

### Question 9. The most frequent word

(2 marks)

Write `most_frequent(counts)` that takes a dictionary of the kind
`word_counts` produces and returns the word with the highest count.

When two or more words tie for the highest count, return whichever of them
comes first alphabetically, so that the answer is always the same. Return
`None` when the dictionary is empty.

Walk over the dictionary with `.items()`. Do not sort the dictionary.

Example:

`most_frequent({'the': 3, 'cat': 2, 'hat': 1})` returns `'the'`

```python
def most_frequent(counts):
    best_word = None
    best_count = 0
    for word, count in counts.items():
        if best_word is None or count > best_count:
            best_word = word
            best_count = count
        elif count == best_count and word < best_word:
            best_word = word
    return best_word
```

```python
# Test cases

from hamcrest import assert_that, equal_to
assert_that(most_frequent({"the": 3, "cat": 2, "hat": 1}), equal_to("the"))
assert_that(most_frequent({"solo": 9}), equal_to("solo"))
assert_that(most_frequent({}), equal_to(None))

from hamcrest import assert_that, equal_to
assert_that(most_frequent({"b": 2, "a": 2}), equal_to("a"))
assert_that(most_frequent({"a": 2, "b": 2}), equal_to("a"))
assert_that(most_frequent({"zebra": 1, "ant": 1, "mouse": 1}), equal_to("ant"))

print('Question 9: all test cases passed')
```

### Question 10. Building a phone book from text lines

(2 marks)

A phone book arrives as a list of strings, one entry per string, with the
name and the number separated by a colon. The spacing around the colon is not
reliable.

Write `parse_phonebook(lines)` that returns a dictionary mapping each name to
its number as an integer. Strip the stray spaces from both halves. You may
assume every line holds exactly one colon.

Return an empty dictionary for an empty list.

Example:

`parse_phonebook(['mom : 1234', 'dad:5678'])` returns
`{'mom': 1234, 'dad': 5678}`

```python
def parse_phonebook(lines):
    book = {}
    for line in lines:
        name, number = line.split(":")
        book[name.strip()] = int(number.strip())
    return book
```

```python
# Test cases

from hamcrest import assert_that, equal_to
assert_that(parse_phonebook(["mom : 1234", "dad:5678"]), equal_to({"mom": 1234, "dad": 5678}))
assert_that(parse_phonebook([]), equal_to({}))

from hamcrest import assert_that, equal_to
assert_that(parse_phonebook(["  work  :  1022  "]), equal_to({"work": 1022}))
assert_that(parse_phonebook(["a:1", "b:2", "c:3"]), equal_to({"a": 1, "b": 2, "c": 3}))

book = parse_phonebook(["mom:1234"])
assert_that(book["mom"], equal_to(1234))
assert_that(isinstance(book["mom"], int), equal_to(True))

print('Question 10: all test cases passed')
```

### Question 11. Saving lines to a file and reading them back

(2 marks)

Write two functions that are the reverse of each other.

`save_lines(path, lines)` takes a list of strings and writes them to the file
at `path`, one string per line. Every line in the file ends with a newline
character, including the last one.

`load_lines(path)` reads that file back and returns the list of strings without
the newline characters.

Open both files with the `with` statement so that they are closed for you. You
may assume the strings hold no leading or trailing spaces of their own.

Example:

after `save_lines("notes.txt", ["alpha", "beta"])` the file holds
`"alpha\nbeta\n"`, and `load_lines("notes.txt")` returns `['alpha', 'beta']`

```python
def save_lines(path, lines):
    with open(path, "w") as f:
        for line in lines:
            f.write(line + "\n")


def load_lines(path):
    with open(path) as f:
        raw_lines = f.readlines()
    result = []
    for line in raw_lines:
        result.append(line.strip())
    return result
```

```python
# Test cases

from hamcrest import assert_that, equal_to
import os
import shutil
import tempfile

folder = tempfile.mkdtemp()
path = os.path.join(folder, "notes.txt")

save_lines(path, ["alpha", "beta", "gamma"])
assert_that(load_lines(path), equal_to(["alpha", "beta", "gamma"]))

with open(path) as f:
    raw = f.read()
assert_that(raw, equal_to("alpha\nbeta\ngamma\n"))
shutil.rmtree(folder)

from hamcrest import assert_that, equal_to
import os
import shutil
import tempfile

folder = tempfile.mkdtemp()

empty = os.path.join(folder, "empty.txt")
save_lines(empty, [])
assert_that(load_lines(empty), equal_to([]))

single = os.path.join(folder, "single.txt")
save_lines(single, ["only one"])
assert_that(load_lines(single), equal_to(["only one"]))
with open(single) as f:
    assert_that(f.read(), equal_to("only one\n"))

shutil.rmtree(folder)

print('Question 11: all test cases passed')
```

### Question 12. Copying every alternate line

(2 marks)

Write `copy_alternate(src_path, dst_path)` that reads the file at `src_path`
and writes every alternate line of it to the file at `dst_path`, starting with
the very first line. So lines 1, 3, 5 and so on are kept and the rest are
dropped.

Return the number of lines written.

Read the whole file with `.readlines()` and write the kept lines with
`.writelines()`. Remember that `.readlines()` leaves the newline character on
the end of each line, which is exactly what `.writelines()` needs, because it
does not add one of its own.

Example:

if the source file holds `one`, `two`, `three`, `four`, `five` on five lines,
then the destination file ends up holding `one`, `three`, `five`, and the
function returns `3`

```python
def copy_alternate(src_path, dst_path):
    with open(src_path) as f:
        lines = f.readlines()
    kept = []
    index = 0
    while index < len(lines):
        kept.append(lines[index])
        index = index + 2
    with open(dst_path, "w") as f:
        f.writelines(kept)
    return len(kept)
```

```python
# Test cases

from hamcrest import assert_that, equal_to
import os
import shutil
import tempfile

folder = tempfile.mkdtemp()
src = os.path.join(folder, "src.txt")
dst = os.path.join(folder, "dst.txt")

with open(src, "w") as f:
    f.write("one\ntwo\nthree\nfour\nfive\n")

assert_that(copy_alternate(src, dst), equal_to(3))
with open(dst) as f:
    assert_that(f.read(), equal_to("one\nthree\nfive\n"))

shutil.rmtree(folder)

from hamcrest import assert_that, equal_to
import os
import shutil
import tempfile

folder = tempfile.mkdtemp()
src = os.path.join(folder, "src.txt")
dst = os.path.join(folder, "dst.txt")

with open(src, "w") as f:
    f.write("")
assert_that(copy_alternate(src, dst), equal_to(0))
with open(dst) as f:
    assert_that(f.read(), equal_to(""))

with open(src, "w") as f:
    f.write("solo\n")
assert_that(copy_alternate(src, dst), equal_to(1))
with open(dst) as f:
    assert_that(f.read(), equal_to("solo\n"))

with open(src, "w") as f:
    f.write("a\nb\nc\nd\n")
assert_that(copy_alternate(src, dst), equal_to(2))
with open(dst) as f:
    assert_that(f.read(), equal_to("a\nc\n"))

shutil.rmtree(folder)

print('Question 12: all test cases passed')
```

### Question 14. Picking rows with a mask

A `DataFrame` called `products` has a column `rating` and a column `price`.

Write `good_and_cheap(products, min_rating, max_price)` that returns the rows
whose rating is greater than or equal to `min_rating` and whose price is less
than or equal to `max_price`. Keep the rows in their original order and leave
`products` itself unchanged.

Build one boolean condition per column and combine them with `&`. Wrap each
condition in its own brackets, because `&` binds more tightly than the
comparison operators do.

Example:

with ratings `[4.5, 3.0, 4.8]` and prices `[200, 150, 900]`, asking for a
rating of at least `4.0` and a price of at most `500` keeps only the first row

```python
import pandas as pd


def good_and_cheap(products, min_rating, max_price):
    return products[(products['rating'] >= min_rating) & (products['price'] <= max_price)]
```

for a `reading` column holding `1.0`, `NaN`, `3.0`, `count_missing` returns `1`
and filling with `0.0` gives `1.0`, `0.0`, `3.0`

```python
import pandas as pd


def count_missing(data, column):
    return data[column].isnull().sum()


def fill_missing(data, column, value):
    filled = data.copy()
    filled[column] = filled[column].fillna(value)
    return filled
```

```python
# Test cases

from hamcrest import assert_that, equal_to
import pandas as pd
import numpy as np

data = pd.DataFrame({'sensor': ['a', 'b', 'c', 'd'],
                     'reading': [1.0, np.nan, 3.0, np.nan]})

assert_that(int(count_missing(data, 'reading')), equal_to(2))
assert_that(int(count_missing(data, 'sensor')), equal_to(0))

from hamcrest import assert_that, equal_to
import pandas as pd
import numpy as np

data = pd.DataFrame({'sensor': ['a', 'b', 'c', 'd'],
                     'reading': [1.0, np.nan, 3.0, np.nan]})
filled = fill_missing(data, 'reading', 0.0)

assert_that(isinstance(filled, pd.DataFrame), equal_to(True))
assert_that(list(filled['reading']), equal_to([1.0, 0.0, 3.0, 0.0]))
assert_that(list(filled['sensor']), equal_to(['a', 'b', 'c', 'd']))
assert_that(int(count_missing(filled, 'reading')), equal_to(0))
assert_that(int(count_missing(data, 'reading')), equal_to(2))

print('Question 15: all test cases passed')
```
