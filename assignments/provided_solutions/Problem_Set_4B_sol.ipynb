```python
import numpy as np
from hamcrest import (
    assert_that, all_of, close_to, contains, equal_to,
    greater_than_or_equal_to, has_length, less_than,
    less_than_or_equal_to, only_contains
)

np.random.seed(123)  # Do not change or repeat this seed.
```

**Important:** The random seed is fixed once in the first code cell. Run that cell before starting, and do not change or reset the seed anywhere else in the notebook.


## Question 1. A four-sided die


Write a function `tet(size=1)` that returns `size` rolls of a fair four-sided die using `np.random.randint`. Every roll must be one of 1, 2, 3 or 4. do not use `np.random.choice`.


```python
def tet(size=1):
    return np.random.randint(low=1, high=5, size=size)
```

```python
rolls = tet(8)
assert_that(rolls.tolist(), contains(3, 2, 3, 3, 1, 3, 3, 2), 'first eight rolls')

more_rolls = tet(100)
assert_that(more_rolls.tolist(), has_length(100), 'number of rolls')
assert_that(more_rolls.tolist(), only_contains(1, 2, 3, 4), 'possible die values')
assert_that(tet(0).tolist(), equal_to([]), 'zero rolls')

print('Question 1. A four-sided die: all test cases passed')
```

## Question 2. Uniform random numbers in an interval


Write `uniform_between(left, right, size=1)` using `np.random.random`. It must return `size` uniformly distributed values in the interval $[left,right)$. Scale and shift the numbers from $[0,1)$; do not use `np.random.uniform`.


```python
def uniform_between(left, right, size=1):
    x = np.random.random(size)
    return left + (right - left)*x
```

```python
values = uniform_between(-2, 3, 5)
assert_that(values.tolist(), has_length(5), 'number of generated values')
assert_that(values.tolist(), only_contains(all_of(greater_than_or_equal_to(-2), less_than(3))), 'values in [-2, 3)')
assert_that(uniform_between(4, 8, 0).tolist(), equal_to([]), 'zero generated values')

print('Question 2. Uniform random numbers in an interval: all test cases passed')
```

## Question 3. Probability from a normal random experiment


Write `normal_interval_probability(mean, std, size, left, right)`. Generate `size` observations using `np.random.normal`. Use a simple `for` loop to count how many observations lie in the closed interval $[left,right]$, and return the relative count. You may assume `size > 0`. Use the fixed seed from the first cell; do not set a seed inside the function.


```python
def normal_interval_probability(mean, std, size, left, right):
    values = np.random.normal(loc=mean, scale=std, size=size)
    count = 0
    for value in values:
        if left <= value and value <= right:
            count = count + 1
    return count/size
```

```python
assert_that(normal_interval_probability(10, 0, 10, 9, 11), equal_to(1.0), 'all observations inside')
assert_that(normal_interval_probability(10, 0, 10, 11, 12), equal_to(0.0), 'all observations outside')

answer = normal_interval_probability(10, 2, 20, 8, 12)
assert_that(answer, all_of(greater_than_or_equal_to(0), less_than_or_equal_to(1)), 'valid probability')

print('Question 3. Probability from a normal random experiment: all test cases passed')
```

## Question 4. Sum of two dice


Write `dice_sum_probabilities(size)`. Generate `size` rolls for each of two fair six-sided dice using `np.random.randint`, and add the two arrays to obtain their sums. Use `np.histogram(sums, bins=range(2, 14), density=True)` to construct the empirical PMF. Return $P(\text{sum}=7)$ and $P(\text{sum}\geq10)$.


```python
def dice_sum_probabilities(size):
    die_1 = np.random.randint(low=1, high=7, size=size)
    die_2 = np.random.randint(low=1, high=7, size=size)
    sums = die_1 + die_2
    pmf, bins = np.histogram(sums, bins=range(2, 14), density=True)
    probability_7 = pmf[5]
    probability_10_or_more = np.sum(pmf[8:])
    return probability_7, probability_10_or_more
```

```python
p7, p10 = dice_sum_probabilities(10000)
assert_that(p7, close_to(1/6, 0.03), 'probability that the sum is 7')
assert_that(p10, close_to(1/6, 0.03), 'probability that the sum is at least 10')
assert_that(p7, all_of(greater_than_or_equal_to(0), less_than_or_equal_to(1)))
assert_that(p10, all_of(greater_than_or_equal_to(0), less_than_or_equal_to(1)))

print('Question 4. Sum of two dice: all test cases passed')
```

## Question 5. A biased coin from uniform random numbers


Write `biased_coin(p, size)` using `np.random.random`. For every generated number smaller than `p`, append the string `"H"` to a result list; otherwise append `"T"`. Return the list after `size` trials. Use a simple `for` loop and do not use `np.random.choice`.


```python
def biased_coin(p, size):
    random_numbers = np.random.random(size)
    results = []
    for number in random_numbers:
        if number < p:
            results.append("H")
        else:
            results.append("T")
    return results
```

```python
assert_that(biased_coin(0, 5), contains('T', 'T', 'T', 'T', 'T'), 'p = 0')
assert_that(biased_coin(1, 5), contains('H', 'H', 'H', 'H', 'H'), 'p = 1')

trials = biased_coin(0.5, 20)
assert_that(trials, has_length(20), 'number of trials')
assert_that(trials, only_contains('H', 'T'), 'possible outcomes')
assert_that(biased_coin(0.4, 0), equal_to([]), 'zero trials')

print('Question 5. A biased coin from uniform random numbers: all test cases passed')
```
