2. Write uniform_between(left, right, size=1) using np.random.random. It must return size uniformly distributed values in the interval [left, right). Scale and shift numbers from [0, 1); do not use np.random.uniform and do not set or change the random seed inside the function.

```python
import numpy as np

def uniform_between(left, right, size=1):
    return left + (right-left) * np.random.random(size)
```

3. Write normal_interval_probability(mean, std, size, left, right). Generate size observations using np.random.normal. Use a simple for loop to count how many observations lie in the closed interval [left, right], and return the relative count. You may assume size > 0. Do not set or change the random seed inside the function.

```python
import numpy as np

def normal_interval_probability(mean, std, size, left, right):
    count=0
    for i in np.random.normal(mean,std,size):
      if i>= left and i<=right:
        count+=1
    return count/size
```
