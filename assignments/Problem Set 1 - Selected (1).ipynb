### Question 3

Create an array $x$ containing 20 evenly spaced values in the range $1 \leq x \leq 20$ using the `linspace()` function. Create a second array containing the **logarithmic values** of the elements of $x$, and a third array containing the **cosine values** of the elements of $x$.

$$
y = \log(x)
$$

$$
z = \cos(x)
$$

Plot the second and third arrays against $x$ on the **same graph**. Use a **solid line with blue circle markers** for the logarithmic values and a **dash-dotted line** for the cosine values. Add a suitable title, label the $x$-axis and $y$-axis, and display a legend.

Finally, save the plot as an image file.

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(1,20,20)
y=np.log(x)
z=np.cos(x)

plt.plot(x,y,"bo-")
plt.plot(x,z,"--")

plt.title("Plot")
plt.xlabel("Numbers")
plt.ylabel("log/cos")
plt.legend(['log(x)',"cos(x)"])

plt.savefig('Plot_q_3.png')
```

### Question 5

Create two lists:


`students = ["Riya", "Arjun", "Meera", "Kabir", "Neha"]`

`marks=[68,91,76,54,83]`


Create an empty list called `grades`.

Using a loop, assign the grades based on the following condition
    
    - "A" if mark ≥ 85
    - "B" if 70 ≤ mark < 85
    - "C" if 50 ≤ mark < 70
    - "F" if mark < 50
    
Using a loop, print each student’s name, mark, and grade.

```python

students = ["Riya", "Arjun", "Meera", "Kabir", "Neha"]
marks=[68,91,76,54,83]
grades=[]

for mark in marks:
    if mark>=85:
        grades.append("A")
    elif mark>=70:
        grades.append("B")
    elif mark>=50:
        grades.append("C")
    else:
        grades.append("F")

for i in range(0,len(students)):
    print(students[i] , " : ", marks[i]," : ", grades[i])
```

### Question 6

A shop stores the prices of different products in a list:

`prices = [120, 450, 75, 800, 250, 60, 950, 180, 320, 500]`

Extract the prices of the **last five products** and store them in a new list called `last_prices`.

Create a new list called `expensive` containing only those prices from `last_prices` that are **greater than 300**.

Finally:

- Print `last_prices`.
- Print the `expensive` list.
- Count and print the number of expensive products.
- Print the highest price among the expensive products.

```python
prices = [120, 450, 75, 800, 250, 60, 950, 180, 320, 500]

last_prices=prices[len(prices)-5:]
expensive=[]

for price in last_prices:
    if price >300:
        expensive.append(price)

print("last_prices : ",last_prices)
print("expensive : ",expensive)
print("Count of expensive : ",len(expensive))
print("Max of expensive : ",max(expensive))
```

### Question 7

### Question 8
The daily temperatures recorded over one week are stored in the list `temperatures = [29, 31, 28, 35, 33, 30, 27]`.

Using a loop, calculate the total and average temperature. Also find how many days had a temperature strictly above the average. Print the total, average, and count. Do not use `sum()`, `max()`, or `min()`; this problem is meant to practise accumulation and comparison in a loop.

```python
temperatures = [29, 31, 28, 35, 33, 30, 27]

total=0
average=0
number_of_days_above_average_temperature=0

for temp in temperatures:
    total+=temp

average=total/len(temperatures)

for temp in temperatures:
    if temp> average:
        number_of_days_above_average_temperature+=1

print("Total : ",total)
print("Average : ",average)
print("Number of days above average temperature : ",number_of_days_above_average_temperature)
```

### Question 10
Write a function `classify_mark(mark)` that returns `"Excellent"` for marks at least 85, `"Good"` for marks from 70 through 84, `"Pass"` for marks from 50 through 69, and `"Fail"` otherwise.

Apply the function in a loop to `marks = [92, 76, 48, 85, 63, 39, 71]`. Store the returned categories in a list called `categories`, and print each mark alongside its category.

```python
def classify_mark(mark):
    if mark>=85:
        return "Excellent"
    elif mark >= 70:
        return "Good"
    elif mark >= 50:
        return "Pass"
    else:
        return "Fail"

marks=[92, 76, 48, 85, 63, 39, 71]
categories=[]
for i in range(0,len(marks)):
    categories.append(classify_mark(marks[i]))
    print("Mark : ", marks[i]," Category : ",categories[i])
```
