## List Rotation

Write `rotate_list(seq, k)` that returns a new list with the elements of `seq` rotated cyclically to the right by `k` positions.

- Positive `k`: rotate right by `k`, wrapping the trailing elements to the front.
- Negative `k`: rotate left by `|k|`, wrapping the leading elements to the back.
- If `|k| >= len(seq)`, use the effective rotation `k mod n`.
- `seq` must not be modified. Return `[]` if `seq` is empty.

**Example:**

`rotate_list([10, 20, 30, 40, 50, 60], 2)` → `[50, 60, 10, 20, 30, 40]`


````python
def rotate_list(seq, k):
    n = len(seq)
    if n == 0:
        return []
    k = k % n
    if k == 0:
        return seq[:]
    return seq[-k:] + seq[:-k]
````

## Fixing Faulty Negative Readings

A soil-moisture sensor sometimes glitches and logs a negative value, which is physically impossible. A simple fix: replace every negative reading with the mean of all the genuinely non-negative readings in the same batch.

Write `replace_negatives_with_mean(arr)` that takes a 1D NumPy array `arr` and returns a **new** array (do not modify `arr` in place) where every negative entry has been replaced by the mean of the non-negative entries of `arr`. Do not use a Python loop; use boolean indexing.

*Hint: if you're not sure `arr` is already a float array, cast it up front with `arr.astype(float)` before doing any of this — otherwise assigning a mean (which is a float) into an integer array will silently truncate it.*

**Example:**

`replace_negatives_with_mean(np.array([4, -1, 6, -3, 8]))` → `array([4., 6., 6., 6., 8.])`

(Non-negative values are `4, 6, 8`, whose mean is `6`.)


````python
def replace_negatives_with_mean(arr):
    arr = np.asarray(arr)
    result = arr.copy().astype(float)
    nonneg_mean = arr[arr >= 0].mean()
    result[result < 0] = nonneg_mean
    return result
````

## Standardizing Exam Scores

A teacher wants to convert a class's raw exam scores into z-scores, so that scores can be compared across different tests. The z-score of a value `x` is `(x - mean) / std`, where `std` is the population (biased) standard deviation of the whole batch (i.e. `ddof=0`, NumPy's default).

Write `z_scores(data)` that takes a 1D array-like `data` of raw scores and returns a NumPy array of the corresponding z-scores.

**Example:**

`z_scores(np.array([2, 4, 4, 4, 5, 5, 7, 9]))` → `array([-1.5, -0.5, -0.5, -0.5, 0., 0., 1., 2.])`


````python
def z_scores(data):
    data = np.asarray(data, dtype=float)
    return (data - data.mean()) / data.std()
````

## Monsoon Rainfall Record

The IMD tracks how rainfall in a city changes from one day to the next. Given this list of day-to-day changes (in mm), how many days saw an increase, decrease, and no change?

`count_signs(changes)` → `[num_positive, num_negative, num_zero]`

**Example:**

`count_signs([4, -6, 0, 0, 9, -2, 5])` → `[3, 2, 2]`


````python
def count_signs(changes):
    pos = neg = zero = 0
    for v in changes:
        if v > 0:
            pos += 1
        elif v < 0:
            neg += 1
        else:
            zero += 1
    return [pos, neg, zero]
````

## Power Grid Status

A substation logs its supply status every minute — `1` for supply on, `0` for load-shedding. The grid engineer isn't interested in the raw log, only in how often the status *flips* from one minute to the next.

Write `count_changes(readings)` to compute this count.

**Example:**

`count_changes([0, 0, 0, 1, 1, 0, 1, 1, 1])` → `3`


````python
def count_changes(readings):
    count = 0
    for i in range(1, len(readings)):
        if readings[i] != readings[i - 1]:
            count += 1
    return count
````

## Matching UPI Transactions

To process a refund, a shopkeeper is going through a list `amounts` of transaction amounts, in **whole rupees** (so every value is an integer - comparing them for exact equality is safe here, unlike with amounts that could have paise/decimal parts).

She wants to know how many **pairs of positions** `(i, j)` in the list, with `i < j` (so each pair of entries is counted once, not twice, and no entry is paired with itself), have `amounts[i] + amounts[j] == target`.

Write `count_pairs(amounts, target)` returning how many such pairs of positions exist.

**Example:**

`count_pairs([1, 5, 3, 5, 2], 6)` → `2`

(The two qualifying pairs of *positions* are `(0, 1)`, since `amounts[0] + amounts[1] = 1 + 5 = 6`, and `(0, 3)`, since `amounts[0] + amounts[3] = 1 + 5 = 6` - position 3 holds a second `5`, distinct from position 1. No other pair of positions sums to 6.)


````python
def count_pairs(amounts, target):
    count = 0
    n = len(amounts)
    for i in range(n):
        for j in range(i + 1, n):
            if amounts[i] + amounts[j] == target:
                count += 1
    return count
````

## Ridge Temperature Swing

A weather station in the Nilgiri hills records temperatures (°C) for several days, with one row per day and one column per hourly reading. For each day, the "swing" is the difference between the hottest and coldest reading recorded that day.

Write `row_range(matrix)` that takes a 2D NumPy array `matrix` (one row per day) and returns a 1D NumPy array containing the swing for each row.

*You're encouraged to do this with NumPy's `axis` argument instead of a Python loop over rows, but a loop-based solution is also fine — there's no separate check for how you got there, only for the returned values.*

**Example:**

```
row_range(np.array([[3, 7, 1], [10, 10, 10], [-2, 5, 0]]))
```
→ `array([6, 0, 7])`


````python
def row_range(matrix):
    matrix = np.asarray(matrix)
    return np.max(matrix, axis=1) - np.min(matrix, axis=1)
````

## Per-Sensor Reading Cleanup

A factory floor has several pressure sensors, each with its **own** valid operating range — they're different makes of sensor, calibrated differently. Readings are logged in a 2D NumPy array `readings` of shape `(num_sensors, num_readings)`, one row per sensor. Each sensor's valid range is given by two 1D arrays `low` and `high`, each of length `num_sensors`, where `low[i]` and `high[i]` are the valid bounds for sensor `i`.

Write `clip_per_sensor(readings, low, high)` that returns a new array of the same shape as `readings`, in which every entry in row `i` has been clipped to `[low[i], high[i]]` — **using broadcasting**, not a Python loop over sensors.

*Hint: `low` and `high` are 1D, one value per row of `readings`. You'll need to turn each into a column before it can broadcast correctly against a 2D array — `low[:, np.newaxis]` does this.*

**Example:**

```python
readings = np.array([[-5,  3, 12],
                      [ 8, -1, 20],
                      [ 2, 25, -10]])
low  = np.array([0, -2, 5])
high = np.array([10, 15, 15])

clip_per_sensor(readings, low, high)
```
→
```python
array([[ 0.,  3., 10.],
       [ 8., -1., 15.],
       [ 5., 15.,  5.]])
```


````python
def clip_per_sensor(readings, low, high):
    readings = np.asarray(readings, dtype=float)
    low = np.asarray(low, dtype=float)[:, np.newaxis]
    high = np.asarray(high, dtype=float)[:, np.newaxis]
    return np.clip(readings, low, high)
````

## Normalizing an Image's Colour Channels

A small colour image is represented as a NumPy array `img` of shape `(H, W, 3)` — height, width, and 3 colour channels (R, G, B). The three channels were captured by sensors with different gains, so their pixel values aren't on comparable scales: one channel might range over `0`–`30`, another over `50`–`200`.

Write `normalize_channels(img)` that returns a new array of the same shape in which **each channel independently** has been rescaled (min-max normalized) so its own minimum value becomes `0.0` and its own maximum value becomes `1.0`. Use broadcasting across the whole image — do not write a `for`/`while` loop or a list comprehension. You may assume no channel is constant (so you won't divide by zero).

**Example:**

```python
img = np.array([[[0, 10, 100], [10, 20, 150]],
                 [[20, 0,  50], [30, 30, 200]]])

normalize_channels(img)
```
→
```python
array([[[0.   , 0.333, 0.333],
        [0.333, 0.667, 0.667]],

       [[0.667, 0.   , 0.   ],
        [1.   , 1.   , 1.   ]]])
```

(Channel 0, the R values `0, 10, 20, 30`, range from 0 to 30, so they map to `0, 0.333, 0.667, 1`. Each channel is rescaled the same way, independently, using its own min and max.)


````python
def normalize_channels(img):
    img = np.asarray(img, dtype=float)
    cmin = img.min(axis=(0, 1))   # shape (3,) -- one min per channel
    cmax = img.max(axis=(0, 1))   # shape (3,) -- one max per channel
    return (img - cmin) / (cmax - cmin)
````

## Estimating a Probability by Simulation

Suppose `X` and `Y` are independent random variables, each uniformly distributed on `(0, 1)`. We want to estimate `P(X + Y > 1)` using Monte Carlo simulation (its true value is `0.5`, but pretend you don't know that).

Complete `estimate_probability(n_trials, seed)`, started for you below — the seed is already set, so you don't need to call `np.random.seed` yourself:

```python
def estimate_probability(n_trials, seed):
    np.random.seed(seed)
    # your code here: generate n_trials samples of X and of Y from
    # np.random.random, and return the fraction of trials with X + Y > 1,
    # as a Python float
```

**Example:**

`estimate_probability(200000, 7)` → approximately `0.5` (e.g. `0.50028`)


````python
def estimate_probability(n_trials, seed):
    np.random.seed(seed)
    x = np.random.random(n_trials)
    y = np.random.random(n_trials)
    return float(np.mean((x + y) > 1))
````

## Estimating Pi by Throwing Darts

A classic Monte Carlo trick: estimate the value of $\pi$ by throwing random points at a square dartboard and seeing how many land inside the inscribed circle.

Complete `estimate_pi(n_points, seed)`, started for you below — the seed is already set, so you don't need to call `np.random.seed` yourself:

```python
def estimate_pi(n_points, seed):
    np.random.seed(seed)
    # your code here: generate n_points random 2D points, with x and y
    # coordinates independently and uniformly distributed in [-1, 1]
    # (np.random.uniform is fine here), compute the fraction that land
    # inside the unit circle (x**2 + y**2 <= 1), and return the pi
    # estimate 4 * that fraction, as a Python float
```

**Example:**

`estimate_pi(500000, 7)` → approximately `3.14159` (e.g. `3.142712`)


````python
def estimate_pi(n_points, seed):
    np.random.seed(seed)
    points = np.random.uniform(-1, 1, size=(n_points, 2))
    inside = (points[:, 0]**2 + points[:, 1]**2) <= 1
    fraction_inside = np.mean(inside)
    return float(fraction_inside * 4)
````

## Flagging Outlier Transactions

A bank wants to flag unusually large or small transaction amounts in a batch, using the standard IQR (interquartile range) rule: compute `Q1` (25th percentile) and `Q3` (75th percentile) of the data, and let `IQR = Q3 - Q1`. Any value below `Q1 - 1.5 * IQR` or above `Q3 + 1.5 * IQR` is considered an outlier.

Write `iqr_outliers(data)` that takes a 1D NumPy array `data` and returns a NumPy array containing just the outlier values, **in their original order** in `data`.

*Hint: to combine the two conditions (too low OR too high) into a single boolean mask, you can use `np.logical_or(condition1, condition2)`, or equivalently the bitwise `|` operator: `condition1 | condition2`.*

**Example:**

`iqr_outliers(np.array([10, 12, 12, 13, 12, 11, 14, 13, 100, 12, -50]))` → `array([100, -50])`


````python
def iqr_outliers(data):
    data = np.asarray(data)
    q1, q3 = np.percentile(data, [25, 75])
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    is_outlier = np.logical_or(data < low, data > high)
    return data[is_outlier]
````
