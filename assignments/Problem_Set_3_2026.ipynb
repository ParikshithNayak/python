## Question 1. Even or odd using a bitwise operator

(1 mark)

Write a function named `check_even_odd` that takes a single integer
argument `number` and returns the string `"even"` or the
string `"odd"`. Both strings are lowercase and are compared
exactly, so the case must match.

You must decide using the bitwise AND operator `&`, not
`%` and not `/`.

For example, `check_even_odd(10)` returns `"even"` and
`check_even_odd(7)` returns `"odd"`.

Your function must also work for negative numbers and for zero.


```python
def check_even_odd(number):
    if number & 1 == 0:
        return "even"
    else:
        return "odd"
```

## Question 2. Reverse a number using a while loop

(1 mark)

Write a function named `reverse_number` that takes a non-negative
integer `n` and returns the integer formed by reversing its digits.

Use a `while` loop together with the floor division operator
`//` and the modulo operator `%`. Do not convert the
number to a string.

For example, `reverse_number(23658)` returns `85632`.

Note that `reverse_number(0)` must return `0`, and
`reverse_number(1200)` must return `21`.


```python
def reverse_number(n):
    reversed_n = 0
    while n > 0:
        digit = n % 10          
        reversed_n = reversed_n * 10 + digit 
        n = n // 10             
    return reversed_n
```

## Question 3. Looping with break and continue

(2 marks)

Write a function named `scan_numbers` that takes two integer
arguments, `limit` and `stop_at`, and returns a list built
by walking over the numbers `1, 2, ..., limit` in order and applying
these rules to each number `i`:


- If `i` equals `stop_at`, stop the loop immediately.
- Otherwise, if `i` is divisible by 5, skip it and add nothing.
- Otherwise, if `i` is even, add the string `"Even"` to the list (case-sensitive, capital E).
- Otherwise, add the integer `i` itself to the list.


Use `break` for rule 1 and `continue` for rule 2.

For example, `scan_numbers(12, 9)` returns
`[1, 'Even', 3, 'Even', 'Even', 7, 'Even']`.

Return the list. Do not print it.


```python
def scan_numbers(limit, stop_at):
    result = []
    for i in range(1, limit + 1):
        if i == stop_at:
            break
        if i % 5 == 0:
            continue
        if i % 2 == 0:
            result.append("Even")
        else:
            result.append(i)
    return result
```

## Question 4. Name binding, aliasing and immutable objects

(3 marks)

This question has two parts. Write both functions in the same answer box.

**Part A.** Write a function `alias_and_copy(values)` that takes a
list. Inside the function:


- Bind the name `alias` to the same list object as `values`.
- Bind the name `copy` to a shallow copy of the list, made with slicing.
- Append the number `99` to `alias`.
- Return the four-item tuple `(values, copy, alias is values, copy is values)`.


Because a list is a mutable object and `alias` is only another name
for the same object, the list passed in by the caller is changed as well.

**Part B.** A tuple is immutable, so it cannot be changed in place. Write a
function `replace_in_tuple(data, index, value)` that returns a
**new** tuple which is the same as `data` except that position
`index` holds `value`. The tuple passed in must not be
modified. Build the new tuple with slicing and concatenation.

For example, `replace_in_tuple((10, 20, 30), 1, 99)` returns
`(10, 99, 30)`.


```python
def alias_and_copy(values):
    alias = values
    copy = values[:]
    alias.append(99)
    return (values, copy, alias is values, copy is values)

def replace_in_tuple(data, index, value):
    return data[:index] + (value,) + data[index + 1:]
```

## Question 5. Mutable default arguments

(2 marks)

Writing `def add_reading(value, readings=[])` is a bug,
because the default list is created once when the function is defined and is then
shared by every call that does not supply a list, so the second call sees the
data left behind by the first.

Write a function `add_reading(value, readings=None)` that appends
`value` to `readings` and returns the list. When the caller
does not supply a list, the function must start from a fresh empty list every
time.

Use `None` as the default and test for it with the identity operator
`is`. `None` is the conventional way for a function to say
that no value was supplied for an argument.

So `add_reading(1)` returns `[1]`, and a second call
`add_reading(2)` must also return a single-item list
`[2]`, not `[1, 2]`. When the caller does supply a list,
append to that list and return it.


```python
def add_reading(value, readings=None):
    if readings is None:
        readings = []
    readings.append(value)
    return readings
```

## Question 6. Positional-only and keyword-only parameters

(2 marks)

Write a function with exactly this signature:

`def evaluate_model(predictions, /, threshold=0.5, *, absolute=False):`

The `/` makes `predictions` positional-only, so it can
never be passed by keyword. The `*` makes `absolute`
keyword-only, so it can never be passed by position.

The function takes a list of numbers and returns a list containing `1`
for each prediction that is greater than or equal to `threshold` and
`0` otherwise. If `absolute` is `True`, compare
the absolute value of the prediction instead of the prediction itself.

For `predictions = [0.2, -0.7, 0.4, 0.9, -0.6]`:


- `evaluate_model(predictions)` returns `[0, 0, 0, 1, 0]`
- `evaluate_model(predictions, 0.3)` returns `[0, 0, 1, 1, 0]`
- `evaluate_model(predictions, absolute=True)` returns `[0, 1, 0, 1, 1]`


Return the list. Do not print it.


```python
def evaluate_model(predictions, /, threshold=0.5, *, absolute=False):
    result = []
    for prediction in predictions:
        if absolute:
            prediction = abs(prediction)
        if prediction >= threshold:
            result.append(1)
        else:
            result.append(0)

    return result
```

## Question 7. Functions as values, and lambda

(2 marks)

A function in Python is an ordinary object. It can be bound to a name, stored
in a list, and passed to another function as an argument.

Write all four of the following in the same answer box.


- `apply_operation(values, operation)`: takes a list of numbers and a
function, applies the function to every element and returns the list of results.
- `square(x)`: returns the square of `x`, defined with
`def`.
- `increment`: a name bound to a **lambda** that returns
`x + 1`.
- `apply_each(values, operations)`: here `operations` is
a list of functions. Return a list holding one result list per function, in the
same order, each one produced by calling `apply_operation`.


For example, with `values = [-2, -1, 0, 1, 2]`,

`apply_each(values, [square, increment])`

returns `[[4, 1, 0, 1, 4], [-1, 0, 1, 2, 3]]`.


```python
def apply_operation(values, operation):
    return [operation(value) for value in values]


def square(x):
    return x * x


# Bind increment to a lambda
increment = lambda x: x + 1


def apply_each(values, operations):
    return [apply_operation(values, operation) for operation in operations]
```

## Question 8. Variable scope and namespaces

(2 marks)

A function introduces a new namespace. A name assigned inside a function is
local to that function and does not disturb a name of the same spelling outside
it. A function can read a name from the enclosing module namespace without
assigning to it.

Write the following in one answer box, in this order.


- At module level, bind `counter = 0`.
- `local_shadow()`: assigns `counter = 99` inside itself
and returns that value. The module-level `counter` must stay
`0`.
- `read_only()`: returns the module-level `counter`
without assigning to it.
- `outer(n)`: defines an inner function `double()` that
returns twice the value of `outer`'s argument `n`, then
returns the result of calling `double()`. The name
`double` must exist only inside `outer`.


```python
counter = 0


def local_shadow():
    counter = 99
    return counter

def read_only():
    return counter


def outer(n):
    def double():
        return 2 * n
    return double()
```

## Question 9. Modules, imports and the script guard

(2 marks)

A module is a namespace. When you write `import math`, the name
`math` is bound to the module object and its contents are reached as
`math.pi`. When you write `from math import gcd`, the name
`gcd` is bound directly in your own namespace.

Write the following in one answer box.


- Import the `math` module with `import math`, and
separately import `gcd` from it with
`from math import gcd`.
- `circle_stats(r)`: returns the tuple
`(area, circumference)` for a circle of radius `r`, using
`math.pi`.
- `gcd_of_list(numbers)`: returns the greatest common divisor of a
non-empty list of positive integers, using the imported `gcd` in a
loop.
- `module_name()`: returns the value of the `__name__`
variable.
- At the end of the file, a guard `if __name__ == "__main__":` whose
body calls `circle_stats(1)`. Nothing outside a function should run
when the file is imported as a module.


```python
import math
from math import gcd


def circle_stats(r):
    area = math.pi * r ** 2
    circumference = 2 * math.pi * r
    return (area, circumference)


def gcd_of_list(numbers):
    result = numbers[0]
    for number in numbers[1:]:
        result = gcd(result, number)
    return result

def module_name():
    return __name__

if __name__ == "__main__":
    circle_stats(1)
```
