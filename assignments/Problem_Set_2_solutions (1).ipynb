## Question 1. Even or odd using a bitwise operator

(1 marks)

Write a function named `check_even_odd` that takes a single integer
argument `number` and returns the string `"even"` or the
string `"odd"`. Both strings are lowercase and are compared
exactly, so the case must match.

You must decide using the bitwise AND operator `&`, not
`%` and not `/`.

For example, `check_even_odd(10)` returns `"even"` and
`check_even_odd(7)` returns `"odd"`.

Your function must also work for negative numbers and for zero.

The portal also checks how you wrote it:

- Use the bitwise AND operator &.
- Do not use the modulo operator %. Use bitwise AND.


```python
def check_even_odd(number):
    if number & 1:
        return "odd"
    return "even"
```

```python


# test case 1
assert check_even_odd(10) == "even"
assert check_even_odd(7) == "odd"
assert check_even_odd(0) == "even"

# test case 2
assert check_even_odd(-5) == "odd"
assert check_even_odd(-8) == "even"
assert check_even_odd(1234567) == "odd"

print('Question 1: all test cases passed')
```

## Question 2. Reverse a number using a while loop

(1 marks)

Write a function named `reverse_number` that takes a non-negative
integer `n` and returns the integer formed by reversing its digits.

Use a `while` loop together with the floor division operator
`//` and the modulo operator `%`. Do not convert the
number to a string.

For example, `reverse_number(23658)` returns `85632`.

Note that `reverse_number(0)` must return `0`, and
`reverse_number(1200)` must return `21`.

The portal also checks how you wrote it:

- Use a while loop.
- Do not convert the number to a string.


```python
def reverse_number(n):
    rev = 0
    while n > 0:
        rev = rev * 10 + n % 10
        n = n // 10
    return rev
```

```python


# test case 1
assert reverse_number(23658) == 85632
assert reverse_number(0) == 0
assert reverse_number(7) == 7

# test case 2
assert reverse_number(1200) == 21
assert reverse_number(1000000) == 1
assert reverse_number(987654321) == 123456789

print('Question 2: all test cases passed')
```

## Question 3. Looping with break and continue

(2 marks)

Write a function named `scan_numbers` that takes two integer
arguments, `limit` and `stop_at`, and returns a list built
by walking over the numbers `1, 2, ..., limit` in order and applying
these rules to each number `i`:


- If `i` equals `stop_at`, stop the loop immediately.
- Otherwise, if `i` is divisible by 5, skip it and add nothing.
- Otherwise, if `i` is even, add the string `"Even"` to the list (case-sensitive, capital E).
- Otherwise, add the integer `i` itself to the list.


Use `break` for rule 1 and `continue` for rule 2.

For example, `scan_numbers(12, 9)` returns
`[1, 'Even', 3, 'Even', 'Even', 7, 'Even']`.

Return the list. Do not print it.

The portal also checks how you wrote it:

- Use a break statement.
- Use a continue statement.


```python
def scan_numbers(limit, stop_at):
    result = []
    for i in range(1, limit + 1):
        if i == stop_at:
            break
        if i % 5 == 0:
            continue
        if i % 2 == 0:
            result.append("Even")
        else:
            result.append(i)
    return result
```

```python


# test case 1
assert scan_numbers(12, 9) == [1, 'Even', 3, 'Even', 'Even', 7, 'Even']
assert scan_numbers(6, 100) == [1, 'Even', 3, 'Even', 'Even']

# test case 2
expected = [1, 'Even', 3, 'Even', 'Even', 7, 'Even', 9, 11, 'Even',
            13, 'Even', 'Even', 17, 'Even', 19, 21, 'Even']
assert scan_numbers(50, 23) == expected
assert scan_numbers(1, 1) == []
assert scan_numbers(0, 3) == []

print('Question 3: all test cases passed')
```

## Question 4. Name binding, aliasing and immutable objects

(3 marks)

This question has two parts. Write both functions in the same answer box.

**Part A.** Write a function `alias_and_copy(values)` that takes a
list. Inside the function:


- Bind the name `alias` to the same list object as `values`.
- Bind the name `copy` to a shallow copy of the list, made with slicing.
- Append the number `99` to `alias`.
- Return the four-item tuple `(values, copy, alias is values, copy is values)`.


Because a list is a mutable object and `alias` is only another name
for the same object, the list passed in by the caller is changed as well.

**Part B.** A tuple is immutable, so it cannot be changed in place. Write a
function `replace_in_tuple(data, index, value)` that returns a
**new** tuple which is the same as `data` except that position
`index` holds `value`. The tuple passed in must not be
modified. Build the new tuple with slicing and concatenation.

For example, `replace_in_tuple((10, 20, 30), 1, 99)` returns
`(10, 99, 30)`.


```python
def alias_and_copy(values):
    alias = values
    copy = values[:]
    alias.append(99)
    return values, copy, alias is values, copy is values


def replace_in_tuple(data, index, value):
    return data[:index] + (value,) + data[index + 1:]
```

```python


# test case 1
original = [1, 2, 3]
returned, copied, same, other = alias_and_copy(original)
assert returned == [1, 2, 3, 99]
assert original == [1, 2, 3, 99]
assert copied == [1, 2, 3]
assert same is True
assert other is False
assert returned is original

# test case 2
t = (10, 20, 30)
assert replace_in_tuple(t, 1, 99) == (10, 99, 30)
assert t == (10, 20, 30)
assert isinstance(replace_in_tuple(t, 0, 0), tuple)
assert replace_in_tuple(('a', 'b', 'c', 'd'), 3, 'z') == ('a', 'b', 'c', 'z')
assert replace_in_tuple((5,), 0, 6) == (6,)

print('Question 4: all test cases passed')
```

## Question 5. Counting with a dictionary

(2 marks)

This question has two parts. Write both functions in the same answer box.

**Part A.** Write a function `tally(items)` that takes a sequence
and returns a dictionary mapping each distinct item to the number of times it
appears. Build the dictionary in a loop and use the `get` method to
supply a default count of zero. Do not use
`collections.Counter`.

For example, `tally(['a', 'b', 'a'])` returns
`{'a': 2, 'b': 1}`.

**Part B.** Write a function `lookup(counts, key)` that returns
the count stored for `key` in the dictionary `counts`, and
returns `0` if the key is absent. Use the containership operator
`in` so that a missing key never raises an error.

The portal also checks how you wrote it:

- Do not use collections.Counter. Build the dictionary yourself.


```python
def tally(items):
    counts = {}
    for item in items:
        counts[item] = counts.get(item, 0) + 1
    return counts


def lookup(counts, key):
    if key in counts:
        return counts[key]
    return 0
```

```python


# test case 1
assert tally(['a', 'b', 'a']) == {'a': 2, 'b': 1}
assert tally([]) == {}
assert tally('mississippi') == {'m': 1, 'i': 4, 's': 4, 'p': 2}

# test case 2
marks = ['A', 'B', 'C', 'A', 'B', 'A', 'F']
counts = tally(marks)
assert counts == {'A': 3, 'B': 2, 'C': 1, 'F': 1}
assert lookup(counts, 'A') == 3
assert lookup(counts, 'D') == 0
assert lookup({}, 'anything') == 0

print('Question 5: all test cases passed')
```

## Question 6. Mutable default arguments and detecting a missing argument

(2 marks)

This question has two parts. Write both functions in the same answer box.

**Part A.** Writing `def add_reading(value, readings=[])` is a bug,
because the default list is created once when the function is defined and is then
shared by every call that does not supply a list.

Write a function `add_reading(value, readings=None)` that appends
`value` to `readings` and returns the list. When the caller
does not supply a list, the function must start from a **fresh empty list**
every time. Use `None` as the default and test for it with the
identity operator `is`.

So `add_reading(1)` returns `[1]`, and a second call
`add_reading(2)` must also return a single-item list
`[2]`, not `[1, 2]`.

**Part B.** Sometimes `None` is itself a valid value, so it cannot
be used to mean "not supplied". The usual solution is a private sentinel object.

Create a module-level sentinel with `MISSING = object()` and write a
function `label(value, unit=MISSING)` that returns:


- `"5 (no unit)"` when `unit` is not supplied at all
- `"5 None"` when the caller explicitly passes `None`
- `"5 kg"` when the caller passes `"kg"`


The returned text is compared exactly, so match the spacing and the
case shown above.

Compare against the sentinel with `is`, never with `==`.

The portal also checks how you wrote it:

- Define a function named add_reading.
- Do not use a mutable default. Use readings=None.
- Use None as the default value of readings.
- Test the default with the identity operator: if readings is None.


```python
MISSING = object()


def add_reading(value, readings=None):
    if readings is None:
        readings = []
    readings.append(value)
    return readings


def label(value, unit=MISSING):
    if unit is MISSING:
        return f"{value} (no unit)"
    return f"{value} {unit}"
```

```python


# test case 1
first = add_reading(1)
second = add_reading(2)
assert first == [1]
assert second == [2]
assert first is not second

existing = [10, 20]
returned = add_reading(30, existing)
assert returned == [10, 20, 30]
assert existing == [10, 20, 30]

# test case 2
assert label(5) == "5 (no unit)"
assert label(5, None) == "5 None"
assert label(5, "kg") == "5 kg"
assert label(0, 0) == "0 0"
assert MISSING is not None

print('Question 6: all test cases passed')
```

## Question 7. Positional-only and keyword-only parameters

(2 marks)

Write a function with exactly this signature:

`def evaluate_model(predictions, /, threshold=0.5, *, absolute=False):`

The `/` makes `predictions` positional-only, so it can
never be passed by keyword. The `*` makes `absolute`
keyword-only, so it can never be passed by position.

The function takes a list of numbers and returns a list containing `1`
for each prediction that is greater than or equal to `threshold` and
`0` otherwise. If `absolute` is `True`, compare
the absolute value of the prediction instead of the prediction itself.

For `predictions = [0.2, -0.7, 0.4, 0.9, -0.6]`:


- `evaluate_model(predictions)` returns `[0, 0, 0, 1, 0]`
- `evaluate_model(predictions, 0.3)` returns `[0, 0, 1, 1, 0]`
- `evaluate_model(predictions, absolute=True)` returns `[0, 1, 0, 1, 1]`


Return the list. Do not print it.


```python
def evaluate_model(predictions, /, threshold=0.5, *, absolute=False):
    result = []
    for prediction in predictions:
        value = abs(prediction) if absolute else prediction
        if value >= threshold:
            result.append(1)
        else:
            result.append(0)
    return result
```

```python


# test case 1
predictions = [0.2, -0.7, 0.4, 0.9, -0.6]
assert evaluate_model(predictions) == [0, 0, 0, 1, 0]
assert evaluate_model(predictions, 0.3) == [0, 0, 1, 1, 0]
assert evaluate_model(predictions, absolute=True) == [0, 1, 0, 1, 1]
assert evaluate_model(predictions, 0.65, absolute=True) == [0, 1, 0, 1, 0]
assert evaluate_model([]) == []

# test case 2
try:
    evaluate_model(predictions=[0.2, 0.9])
    raise AssertionError("predictions must be positional-only, so use / in the signature")
except TypeError:
    pass

# test case 3
try:
    evaluate_model([0.2, 0.9], 0.5, True)
    raise AssertionError("absolute must be keyword-only, so use * in the signature")
except TypeError:
    pass

# test case 4
import inspect
params = inspect.signature(evaluate_model).parameters
assert params['predictions'].kind is inspect.Parameter.POSITIONAL_ONLY
assert params['threshold'].kind is inspect.Parameter.POSITIONAL_OR_KEYWORD
assert params['threshold'].default == 0.5
assert params['absolute'].kind is inspect.Parameter.KEYWORD_ONLY
assert params['absolute'].default is False

print('Question 7: all test cases passed')
```

## Question 8. Functions as values, and lambda

(2 marks)

A function in Python is an ordinary object. It can be bound to a name, stored
in a list or a dictionary, and passed to another function as an argument.

Write all four of the following in the same answer box.


- `apply_operation(values, operation)`: takes a list of numbers and a
function, applies the function to every element and returns the list of results.
- `square(x)`: returns the square of `x`, defined with
`def`.
- `increment`: a name bound to a **lambda** that returns
`x + 1`.
- `apply_all(values, operations)`: here `operations` is a
dictionary mapping a name to a function. Return a dictionary with the same keys,
where each value is the result of applying that function to `values`
through `apply_operation`.


For example, with `values = [-2, -1, 0, 1, 2]`,

`apply_all(values, {'square': square, 'increment': increment})`

returns `{'square': [4, 1, 0, 1, 4], 'increment': [-1, 0, 1, 2, 3]}`.


```python
def apply_operation(values, operation):
    result = []
    for value in values:
        result.append(operation(value))
    return result


def square(x):
    return x ** 2


increment = lambda x: x + 1


def apply_all(values, operations):
    result = {}
    for name, operation in operations.items():
        result[name] = apply_operation(values, operation)
    return result
```

## Question 9. Variable scope and namespaces

(2 marks)

A function introduces a new namespace. A name assigned inside a function is
local to that function and does not disturb a name of the same spelling outside
it, unless the function declares that name `global`.

Write the following in one answer box, in this order.


- At module level, bind `counter = 0`.
- `local_shadow()`: assigns `counter = 99` inside itself
and returns that value. The module-level `counter` must stay
`0`.
- `read_only()`: returns the module-level `counter`
without assigning to it.
- `global_update(step)`: declares `counter` as
`global`, increases it by `step`, and returns the new
value.
- `outer(n)`: defines an inner function `double()` that
returns twice the value of `outer`'s argument `n`, then
returns the result of calling `double()`. The name
`double` must exist only inside `outer`.


```python
counter = 0


def local_shadow():
    counter = 99
    return counter


def read_only():
    return counter


def global_update(step):
    global counter
    counter = counter + step
    return counter


def outer(n):
    def double():
        return 2 * n
    return double()
```

```python


# test case 1
before = read_only()
assert local_shadow() == 99
assert read_only() == before, 'local_shadow must not change the module-level counter'
assert counter == before

# test case 2
start = read_only()
assert global_update(5) == start + 5
assert counter == start + 5
assert read_only() == start + 5
assert global_update(3) == start + 8
assert counter == start + 8

# test case 3
import types
assert outer(4) == 8
assert outer(-3) == -6
inner = [c for c in outer.__code__.co_consts
         if isinstance(c, types.CodeType) and c.co_name == 'double']
assert len(inner) == 1, 'double must be defined inside outer'

print('Question 9: all test cases passed')
```

## Question 10. Modules, imports and the script guard

(2 marks)

A module is a namespace. When you write `import math`, the name
`math` is bound to the module object and its contents are reached as
`math.pi`. When you write `from math import gcd`, the name
`gcd` is bound directly in your own namespace.

Write the following in one answer box.


- Import the `math` module with `import math`, and
separately import `gcd` from it with
`from math import gcd`.
- `circle_stats(r)`: returns the tuple
`(area, circumference)` for a circle of radius `r`, using
`math.pi`.
- `gcd_of_list(numbers)`: returns the greatest common divisor of a
non-empty list of positive integers, using the imported `gcd` in a
loop.
- `module_name()`: returns the value of the `__name__`
variable.
- At the end of the file, a guard `if __name__ == "__main__":` whose
body calls `circle_stats(1)`. Nothing outside a function should run
when the file is imported as a module.

The portal also checks how you wrote it:

- Import the math module with: import math
- Also import gcd with: from math import gcd


```python
import math
from math import gcd


def circle_stats(r):
    return math.pi * r * r, 2 * math.pi * r


def gcd_of_list(numbers):
    result = numbers[0]
    for number in numbers[1:]:
        result = gcd(result, number)
    return result


def module_name():
    return __name__


if __name__ == "__main__":
    print(circle_stats(1))
```

```python


# test case 1
import math
area, circumference = circle_stats(2)
assert abs(area - math.pi * 4) < 1e-9
assert abs(circumference - 2 * math.pi * 2) < 1e-9
area, circumference = circle_stats(0)
assert area == 0
assert circumference == 0

# test case 2
assert gcd_of_list([12, 18]) == 6
assert gcd_of_list([24, 36, 60]) == 12
assert gcd_of_list([7]) == 7
assert gcd_of_list([13, 17]) == 1
assert isinstance(module_name(), str)

print('Question 10: all test cases passed')
```
